```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

df=pd.read_csv('/Users/davidferreira/feup/3ano/2Semestre/IA/project2_IA/Assignment2/glioma+grading+clinical+and+mutation+features+dataset-2/TCGA_InfoWithGrade.csv')
features=['Gender','Age_at_diagnosis','Race','IDH1','TP53','ATRX','PTEN','EGFR','CIC','MUC16','PIK3CA','NF1','PIK3R1','FUBP1','RB1','NOTCH1','BCOR','CSMD3','SMARCA4','GRIN2A','IDH2','FAT4','PDGFRA']

target='Grade'

X_train, X_test, y_train, y_test = train_test_split(df[features], df[target], test_size=0.3, random_state=42)

# Cria o modelo de Árvore de Decisão
dt = DecisionTreeClassifier()

# Treina o modelo com os dados de treinamento
dt.fit(X_train, y_train)

# Faz previsões com os dados de teste
y_pred = dt.predict(X_test)

# Avalia a precisão do modelo
accuracy = accuracy_score(y_test, y_pred)
print("A precisão do modelo é:", accuracy)
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

df=pd.read_csv('/Users/davidferreira/feup/3ano/2Semestre/IA/project2_IA/Assignment2/glioma+grading+clinical+and+mutation+features+dataset-2/TCGA_InfoWithGrade.csv')
df['faixa_etaria'] = pd.cut(df['Age_at_diagnosis'], bins=[0, 20, 70, df['Age_at_diagnosis'].max()], labels=['jovem', 'adulto', 'idoso'])

# Define as categorias da variável categórica
categories = ['jovem', 'adulto', 'idoso','desconhecido']

# Cria o dicionário de correspondência
correspondencia = {'jovem': 0, 'adulto': 1, 'idoso': 2, 'desconhecido': -1}

df['faixa_etaria'] = pd.Categorical(df['faixa_etaria'], categories=categories)
df['faixa_etaria'] = df['faixa_etaria'].fillna('desconhecido')
df['faixa_etaria'] = df['faixa_etaria'].map(correspondencia)
df = df.drop('Age_at_diagnosis', axis=1)


# Salvar o DataFrame em um novo arquivo CSV
#df.to_csv('/Users/davidferreira/feup/3ano/2Semestre/IA/project2_IA/Assignment2/glioma+grading+clinical+and+mutation+features+dataset-2/TCGA_InfoWithGrade.csv', index=False)
features=['Gender','Race','IDH1','TP53','ATRX','PTEN','EGFR','CIC','MUC16','PIK3CA','NF1','PIK3R1','FUBP1','RB1','NOTCH1','BCOR','CSMD3','SMARCA4','GRIN2A','IDH2','FAT4','PDGFRA','faixa_etaria']

target='Grade'

X_train, X_test, y_train, y_test = train_test_split(df[features], df[target], test_size=0.3, random_state=42)


# Cria o modelo de Árvore de Decisão
dt = DecisionTreeClassifier()

# Treina o modelo com os dados de treinamento
dt.fit(X_train, y_train)

# Faz previsões com os dados de teste
y_pred = dt.predict(X_test)

# Avalia a precisão do modelo
accuracy = accuracy_score(y_test, y_pred)
print("A precisão do modelo é:", accuracy)
```

    Correlação positiva ou negativa: Os valores na matriz de correlação indicam se existe uma correlação positiva ou negativa entre os atributos. Um valor próximo a 1 indica uma correlação positiva forte, o que significa que os atributos aumentam ou diminuem juntos. Por outro lado, um valor próximo a -1 indica uma correlação negativa forte, ou seja, os atributos têm uma relação inversa.

    Intensidade da correlação: A magnitude dos valores indica a força da correlação. Valores próximos a 1 ou -1 indicam uma correlação forte, enquanto valores próximos a 0 indicam uma correlação fraca.

    Identificação de atributos relacionados: A matriz de correlação permite identificar quais atributos estão mais relacionados entre si. Valores mais altos na matriz indicam uma correlação mais forte, enquanto valores próximos a zero indicam pouca ou nenhuma correlação.

Com base nessas informações, você pode realizar várias experiências, incluindo:

    Seleção de atributos: Se houver atributos altamente correlacionados, pode ser possível remover um deles sem perda significativa de informações. Isso pode ajudar a reduzir a dimensionalidade do conjunto de dados e melhorar o desempenho dos modelos de machine learning.

    Análise de causa e efeito: Se houver uma correlação forte entre dois atributos, isso pode indicar uma relação causal entre eles. No entanto, é importante lembrar que correlação não implica causalidade. É necessário realizar estudos mais aprofundados para determinar a natureza dessa relação.

    Identificação de padrões: A matriz de correlação pode ajudar a identificar padrões interessantes nos dados. Por exemplo, se houver um grupo de atributos fortemente correlacionados, isso pode indicar uma relação intrínseca entre eles, o que pode ser explorado em análises futuras.
