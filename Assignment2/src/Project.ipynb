```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

df=pd.read_csv('/Users/davidferreira/feup/3ano/2Semestre/IA/project2_IA/Assignment2/glioma+grading+clinical+and+mutation+features+dataset/glioma+grading+clinical+and+mutation+features+dataset/Graphics/TCGA_InfoWithGrade.csv')
features=['Gender','Age_at_diagnosis','Race','IDH1','TP53','ATRX','PTEN','EGFR','CIC','MUC16','PIK3CA','NF1', 'PIK3R1', 'FUBP1', 'RB1', 'NOTCH1', 'BCOR', 'CSMD3', 'SMARCA4', 'GRIN2A', 'IDH2', 'FAT4', 'PDGFRA']

target='Grade'

X_train, X_test, y_train, y_test = train_test_split(df[features], df[target], test_size=0.3, random_state=42)

# Cria o modelo de Árvore de Decisão
dt = DecisionTreeClassifier()

# Treina o modelo com os dados de treinamento
dt.fit(X_train, y_train)

# Faz previsões com os dados de teste
y_pred = dt.predict(X_test)

# Avalia a precisão do modelo
accuracy = accuracy_score(y_test, y_pred)
print("A precisão do modelo é:", accuracy)
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

df=pd.read_csv('/Users/davidferreira/feup/3ano/2Semestre/IA/project2_IA/Assignment2/glioma+grading+clinical+and+mutation+features+dataset/glioma+grading+clinical+and+mutation+features+dataset/Graphics/TCGA_InfoWithGrade.csv')
features=['Gender','Age_at_diagnosis', 'NOTCH1']

target='Grade'

X_train, X_test, y_train, y_test = train_test_split(df[features], df[target], test_size=0.3, random_state=42)

# Cria o modelo de Árvore de Decisão
dt = DecisionTreeClassifier()

# Treina o modelo com os dados de treinamento
dt.fit(X_train, y_train)

# Faz previsões com os dados de teste
y_pred = dt.predict(X_test)

# Avalia a precisão do modelo
accuracy = accuracy_score(y_test, y_pred)
print("A precisão do modelo é:", accuracy)
```
