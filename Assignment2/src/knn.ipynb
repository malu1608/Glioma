```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import pandas as pd

dir = r'C:\Users\Utilizador\Documents\3_Ano\2S\IA\IA_2\project2_IA\Assignment2\dataset\TCGA_InfoWithGrade.csv'
data = pd.read_csv(dir)

# Resto do código...

# Step 1: Load your dataset and split it into training and testing sets
# X_train, X_test: feature matrices
# y_train, y_test: target vectors


features=['Gender','Age_at_diagnosis','Race','IDH1','TP53','ATRX','PTEN','EGFR','CIC','MUC16','PIK3CA','NF1','PIK3R1','FUBP1','RB1','NOTCH1','BCOR','CSMD3','SMARCA4','GRIN2A','IDH2','FAT4','PDGFRA']
target = 'Grade'
X_train, X_test, y_train, y_test = train_test_split(data[features], data[target], test_size=0.2, random_state=42)

# Step 2: Preprocess your data if necessary

# Step 3: Choose the value for 'k'
k = 5

# Step 4: Create a k-NN classifier and fit it to the training data
knn = KNeighborsClassifier(n_neighbors=k)
knn.fit(X_train, y_train)

# Step 5: Predict the class labels for the test set
y_pred = knn.predict(X_test)

# Step 6: Evaluate the accuracy of the model
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```
